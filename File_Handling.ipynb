```python
f=open("Q2","a")
n=int(input("Enter the number of students="))
for i in range(n):
    name=input("Enter student name=")
    roll=input("Enter roll number=")
    marks=input("Enter marks=")

    f.write("Name="+name+"\n")
    f.write("Roll No="+roll+"\n")
    f.write("Marks="+marks+"\n")
    f.write("----------------------\n")
f.close()
print("Student records saved successfully.")
```

```python
roll=input("Enter roll number to search=")

f=open("Q2","r")
data=f.readlines()
f.close()
found=False
for i in range(len(data)):
    if data[i].startswith("Roll No="+roll):
        print(data[i-1],end="")
        print(data[i],end="")
        print(data[i+1],end="")
        found=True
        break

if found==False:
    print("Student record not found.")
```

```python
roll=input("Enter roll number to change=")
f=open("Q2","r")
data=f.readlines()
f.close()
found=False
for i in range(len(data)):
    if data[i].startswith("Roll No="+roll):
        new_marks=input("Enter new marks=")
        data[i+1]="Marks="+new_marks+"\n"
        found=True
        break
if found:
    f=open("students.txt","w")
    f.writelines(data)
    f.close()
    print("Marks updated successfully.")
else:
    print("Student record not found.")
```

```python
roll=input("Enter roll number to delete=")
f=open("Q2","r")
data=f.readlines()
f.close()
new_data=[]
i=0
found=False
while i<len(data):
    if data[i+1].startswith("Roll No="+roll):
        found=True
        i=i+4
    else:
        new_data.append(data[i])
        new_data.append(data[i+1])
        new_data.append(data[i+2])
        new_data.append(data[i+3])
        i = i + 4
f=open("students.txt","w")
f.writelines(new_data)
f.close()
if found:
    print("Student record deleted successfully.")
else:
    print("Student record not found.")
```

```python
f=open("Q2","r")
data=f.readlines()
f.close()
students=[]
i=0
while i<len(data):
    roll=data[i].split("=")[1].strip()
    name=data[i+1].split("=")[1].strip()
    marks=int(data[i+2].split("=")[1].strip())
    students.append([roll,name,marks])
    i=i+4
students.sort(key=lambda x:x[2],reverse=True)
print("Roll_No      Name     Marks")
for student in students:
    print(student[0], "     ",student[1], "     ",student[2])
```

```python
f=open("Q2","r")
data=f.readlines()
f.close()
more90=0
less35=0
i=0
while i<len(data):
    marks=int(data[i+2].split("=")[1].strip())
    if marks>90:
        more90=more90+1
    if marks<35:
        less35=less35+1
    i=i+4

print("Students scoring more than 90 =",more90)
print("Students scoring less than 35 =",less35)
```

```python
while True:
    print("\n1.Add")
    print("2.View")
    print("3.Search")
    print("4.Update")
    print("5.Delete")
    print("6.Exit")
    choice=int(input("Enter Choice="))
    
    if choice==1:
        f=open("student.txt","a")
        name=input("Enter Name=")
        roll=input("Enter Roll=")
        marks=input("Enter Marks=")
        f.write(name+","+roll+","+marks+"\n")
        f.close()
        print("Record Added.")
    elif choice==2:
        f=open("student.txt","r")
        for line in f:
            print(line,end="")
        f.close()

    elif choice==3:
        
        roll=input("Enter Roll Number=")
        f=open("student.txt","r")
        found=False
        for line in f:
            data=line.strip().split(",")
            if data[1]==roll:
                print("Name=",data[0])
                print("Roll=",data[1])
                print("Marks=",data[2])
                found=True
        if found==False:
            print("Record Not Found")
        f.close()

    elif choice==4:

        roll=input("Enter Roll Number=")
        f=open("student.txt","r")
        records=[]

        for line in f:
            data=line.strip().split(",")
            if data[1]==roll:
                data[2]=input("Enter New Marks=")
            records.append(",".join(data))

        f.close()

        f=open("student.txt","w")
        for record in records:
            f.write(record+"\n")

        f.close()
        print("Record Updated.")

    elif choice==5:

        value=input("Enter Name or Roll Number=")

        f=open("student.txt","r")
        records=[]

        for line in f:
            data=line.strip().split(",")
            if data[0]!=value and data[1]!=value:
                records.append(line)

        f.close()

        f=open("student.txt","w")

        for record in records:
            f.write(record)

        f.close()

        print("Record Deleted.")

    elif choice==6:
        print("Thank You!")
        break
        
    else:
        print("Invalid Choice")
```
